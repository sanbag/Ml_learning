```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import  FuncAnimation
```

```python
positive_points = np.array([(-2,3),(0,1),(2,-1)])
negatiive_points =  np.array([(-2,1),(0,-1),(2,-3)])
```

```python
plt.figure(figsize=(10,6))
plt.scatter(positive_points[:,0],positive_points[:,1],color='blue',marker='o',label='possitive class')
plt.scatter(negatiive_points[:,0],negatiive_points[:,1],color='red',marker='x',label='negative class')
plt.legend(loc='best')
plt.xlabel('x1')
plt.ylabel('x2')
# plt.grid()
```

```python
x= np.vstack((positive_points,negatiive_points))
y =np.array([1]*len(positive_points)+[-1]*len(negatiive_points))
```

```python
def perceptron(x,y,steps =10):
    theta = np.zeros(x.shape[1])
    theta0 =0
    for t in range(steps):
        for i,xi in enumerate(x):
            if y[i] *(np.dot(xi,theta)+theta0)<=0:
                theta +=y[i]*xi
                theta0 +=y[i]
                break
        plot_decision_boundry(theta,theta0,t+1)
    return theta,theta0
```

```python
def plot_decision_boundry(theta,theta0,iteration):
    plt.figure(figsize=(10,6))
    plt.scatter(positive_points[:,0],positive_points[:,1],color='blue',marker='o',label='possitive class')
    plt.scatter(negatiive_points[:,0],negatiive_points[:,1],color='red',marker='x',label='negative class')

    x_values = np.linspace(-3,3,2)
    if theta[1] !=0:
        y_values = -(theta[0]*x_values + theta0)/theta[1]
        plt.plot(x_values,y_values,label = f'iteration{iteration}')

    else:
        x_val = -theta0/theta[0]
        plt.axvline(x=x_val,label = f'iteration{iteration}')
    
    plt.xlim(-3,3)
    plt.ylim(-4,4)
    plt.xlabel('x1')
    plt.ylabel('x2')
    plt.legend()
    plt.grid(True)
    plt.show()
    
```

```python
theta_final,theta0_final =perceptron(x,y,10)
```
