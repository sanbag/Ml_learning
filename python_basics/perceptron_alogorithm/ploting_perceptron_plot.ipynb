```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
points = [(-2, 3), (-2, 1), (0, 1), (0, -1), (2, -1), (2, -3)]
print(*points)
x_coords,y_coords = zip(*points)
```

```python
# Define the line equation: -2x1 + 3x2 + 1 = 0
# Rearrange to x2 = (2x1 - 1) / 3
x_line = np.linspace(-3,3,100)
y_line = (2 * x_line - 1) / 3
```

```python
plt.scatter(x_coords, y_coords, color='red', label='Points')
plt.plot(x_line, y_line, color='blue', label='Line: -2x1 + 3x2 + 1 = 0')
for x, y in points:
    plt.text(x, y, f'({x},{y})', fontsize=8, ha='right', color='green')

# Add the line name on the line
mid_x = (x_line[0] + x_line[-1]) / 2  # Midpoint of x values
mid_y = (y_line[0] + y_line[-1]) / 2  # Midpoint of y values
plt.text(mid_x, mid_y, 'l1', fontsize=10, color='blue', weight='bold', backgroundcolor='white')


# Define the second line equation: 2x2 + 2 = 0
# Rearrange to x2 = -1
x_line2 =np.linspace(-3,3,100)
y_line2 = -1 * np.ones_like(x_line2)
mid_x2 = (x_line2[0] + x_line2[-1]) / 2
mid_y2 = y_line2[0]  # Constant value for l2
plt.plot(x_line2, y_line2, color='black', linestyle='--', label='Line: l2')
plt.text(mid_x2, mid_y2, 'l2', fontsize=10, color='black', weight='bold', backgroundcolor='white')


# define the third line 2x1 + x2 + 1 = 0
# x2 = -1 - 2x1

x_line3 = np.linspace(-3,3,100)
y_line3 = (-1 -2 * x_line3)
plt.plot(x_line3,y_line3,color = 'grey',linestyle='--')

mid_x3 = x_line3[10]
mid_y3 = y_line3[10]
plt.text(mid_x3,mid_y3,'l3',fontsize=10, color='grey', weight='bold', backgroundcolor='white')
```
